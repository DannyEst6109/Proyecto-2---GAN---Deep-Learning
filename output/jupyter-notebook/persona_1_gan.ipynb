## Comprobar la entrega del dataset

Persona 2 entregará datos y manifiesto documentados según `docs/entrega_dataset.md`. Configurar la ruta y el hash del manifiesto antes de entrenar. Esta comprobación de archivos no verifica licencia, diversidad ni calidad visual.


```python
dataset_path = CONFIG["dataset_path"]
if dataset_path is None:
    print("PENDIENTE: dataset de persona 2. No iniciar entrenamiento.")
else:
    folder = Path(dataset_path)
    if not folder.is_dir():
        raise FileNotFoundError(folder)
    images = [p for p in folder.rglob("*") if p.suffix.lower() in {".png", ".jpg", ".jpeg", ".webp"}]
    if not images:
        raise ValueError("La carpeta no contiene imágenes compatibles.")
    if not CONFIG["dataset_manifest_sha256"]:
        raise ValueError("Falta registrar el hash del manifiesto del dataset.")
    print(f"Archivos de imagen encontrados: {len(images)}")
```
